```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns

df = pd.read_csv('gan_data.csv')
plt.figure(figsize=(10, 6))

# 绘制折线图
columns_to_plot = ['LOSS_G', 'LOSS_G_IDENTITY', 'LOSS_G_GAN', 'LOSS_G_CYCLE','LOSS_D']

# 绘制折线图并设置图表属性
plt.figure(figsize=(10, 6))
sns.set_style('ticks')
sns.set_palette('colorblind')

# 定义要使用的线型和颜色
line_styles = ['-', '--', ':', '-.', '-']
colors = ['blue', 'green', 'red', 'purple', 'orange']


# 绘制折线图并设置图表属性
plt.figure(figsize=(10, 6))

for i, column in enumerate(columns_to_plot):
    plt.plot(df[column], label=column, color=colors[i], linestyle=line_styles[i])

# 设置标题、x轴标签、y轴标签和图例
plt.title('GAN_Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(loc='best')


# 美化坐标轴
sns.despine()
# 显示图表
plt.show()
```
